```python
import pandas as pd

# 1. Import dataset
df = pd.read_csv("dataset.csv")  # Replace with your CSV filename

# 2. Inspect dataset
print("First 5 rows:\n", df.head())
print("\nDataset shape:", df.shape)
print("\nDataset information:")
df.info()
print("\nStatistical summary:\n", df.describe(include="all"))

# 3. Check missing values
print("\nMissing values:\n", df.isnull().sum())

# 4. Check duplicate records
print("\nDuplicate records:", df.duplicated().sum())

# 5. Remove duplicate records
df.drop_duplicates(inplace=True)

# 6. Handle missing values
for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].median())
        else:
            df[col] = df[col].fillna(
                df[col].mode()[0] if not df[col].mode().empty else "Unknown"
            )

# 7. Clean inconsistent text entries
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

# 8. Convert data types (if needed)
# Example:
# df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
# df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# 9. Check cleaned dataset
print("\nMissing values after cleaning:\n", df.isnull().sum())
print("\nDuplicates after cleaning:", df.duplicated().sum())
print("\nCleaned dataset shape:", df.shape)
print("\nCleaned data:\n", df.head())

# 10. Save cleaned dataset (Bonus)
df.to_csv("cleaned_data.csv", index=False)
print("\nCleaned dataset saved as cleaned_data.csv")
```
